```python
import pandas as pd

df=pd.read_csv('../data/processed/recipes.csv')
```

```python
print("All recipes under 30 minutes total time with avg_rating above 4.5: \n")
df.loc[(df['minutes'] < 30) & (df['avg_rating'] > 4.5), ['name','minutes','avg_rating']]
```

```python
print("All desserts with more than 10 ingredients that STILL have avg_rating above 4.0: \n")
df.loc[(df['category'] == 'desserts') & (df['n_ingredients'] > 10) & (df['avg_rating'] > 4.0), ['name','category','n_ingredients','avg_rating']]
```

```python
print("All recipes below their category's median avg_rating: \n")

avg_rate_per_cat=df.groupby('category')['avg_rating'].transform('median')

df.loc[df['avg_rating'] < avg_rate_per_cat, ['name','category','avg_rating']]
```

```python
print("All recipes that are BOTH quick (under 20 minutes) AND simple (5 or fewer ingredients): \n")
df.loc[(df['minutes'] < 20) & (df['n_ingredients'] <= 5), ['name','minutes','n_ingredients']]
```
