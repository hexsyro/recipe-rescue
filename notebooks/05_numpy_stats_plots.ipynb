```python
import pandas as pd
import numpy as np

df=pd.read_csv('../data/processed/recipes.csv')
```

```python
df=df.dropna()
np.corrcoef([df['n_ingredients'], df['n_steps'],df['minutes'], df['avg_rating']])
```

```python
df['rating_z']=df.groupby('category')['avg_rating'].transform(lambda s : (s-s.mean())/s.std())
df.groupby('category')['rating_z'].mean()
```

```python
rated=df[df['rating_count']>=3]
q25=rated['rating_count'].quantile(0.25)
gems=rated.loc[(rated['rating_z']>1.5)&(rated['rating_count'] < q25)]
print(q25, len(gems))
```

```python
df['time']=np.where(df['minutes'] <=10,'quick',
    np.where(df['minutes'] <=30,'medium',
    np.where(df['minutes'] <=120,'long', 'too long')))
df
```
