```python
import pandas as pd
import numpy as np
import math
import ast

df=pd.read_csv('../data/raw/Raw_recipes.csv')


def count_ingredients(df):
    my_ingredients_count=df['ingredients'].apply(ast.literal_eval).apply(len)
    mismatch=df[my_ingredients_count != df['n_ingredients']]

    print(my_ingredients_count)
    print(df['n_ingredients'])
    print(f"mismatched ingredients count; {mismatch}")


count_ingredients(df)
```

```python
MAX_MINUTES=1440

def clean_minutes(df):  
    values=df['minutes'].to_list()
    for raw_value in values:
        # missing values
        if raw_value is None:
            return np.nan
    
        # booleans are ints in Python, but True/False is never a duration
        if isinstance(raw_value, (bool, np.bool_)):
            return np.nan
    
        # make it a float; strings like "45" or " 45.0 " are accepted
        try:
            minutes = float(str(raw_value).strip().replace(",", ""))
        except ValueError:
            return np.nan
    
        # NaN and inf
        if not math.isfinite(minutes):
            return np.nan
    
        minutes = int(round(minutes))
    
        # implausible values
        if minutes <= 0 or minutes > MAX_MINUTES:
            return np.nan
    
        print(minutes)
    
    
clean_minutes(df)
```

```python
NUTRITION_COLS=["calories", "fat_pdv", "sugar_pdv", "sodium_pdv", "protein_pdv",
   "sat_fat_pdv", "carbs_pdv"]

def parse_nutrition(df):  
    df[NUTRITION_COLS] = pd.DataFrame(df['nutrition'].apply(ast.literal_eval).to_list(), columns=NUTRITION_COLS, index=df.index)
    return df
    
    
parse_nutrition(df)
```

```python
import ast

CATEGORIES = ['desserts', 'main-dish', 'breakfast', 'appetizers',
              'side-dishes', 'lunch', 'snacks', 'beverages']

def pick_category(tags):
    for category in CATEGORIES:
        if category in tags:
            return category       # first match wins
    return 'other'                # nothing matched

def assign_category(df):
    tags = df['tags'].apply(ast.literal_eval)
    df['category'] = tags.apply(pick_category)
    return df

df = assign_category(df)
df.merge(df[['id', 'category']], on='id', how='left')
print(df['category'].value_counts())
```
