```python
import pandas as pd

df=pd.read_csv('../data/processed/recipes.csv')
```

```python
ingredient_bucket=pd.cut(df['n_ingredients'], bins=[0,5,10,20,30], labels=['1-5','6-10','11-20','21-30'])
df.pivot_table(index='category',columns=ingredient_bucket,values='avg_rating',aggfunc='mean')
```

```python
time_bucket=pd.cut(df['minutes'], bins=[0,5,15,30,60], labels=['1-5 min','6-15 min','16-30 min','31-60 min']).dropna()
time_pivote=df.pivot_table(index='category',columns=time_bucket,aggfunc='size').reset_index()
time_pivote.melt(id_vars='category', var_name='minutes', value_name='size')
```
